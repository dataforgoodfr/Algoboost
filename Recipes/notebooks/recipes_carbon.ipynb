```python
# import libraries
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
# input data to be displayed

values_rillettes = [2.207, 5.401]
values_gaspacho = [0.808, 1.087]
values_sandwich = [1.181, 2.014]
values_brownie = [46.279, 62.934]

labels_rillettes = ["Rillettes de la mer", "Rillettes au saumon"]
labels_gaspacho = ["Gaspacho pastèque<br>et sauce au kombu royal", "Gaspacho orange<br>et chips de bacon"]
labels_sandwich = ["Sandwich de la mer", "Sandwich thon mayo"]
labels_brownie = ["Brownie à la dulse", "Brownie classique"]

colors = ["#1D3B6E", "#F6A01E"]
```

```python
fig = make_subplots(
    rows=4,
    cols=1,
    vertical_spacing=0.1
)

fig.add_trace(
    go.Bar(
        x=values_rillettes,
        y=labels_rillettes,
        marker_color=colors,
        orientation="h",
        hoverinfo="skip"
    ),
    row=1,
    col=1
)
fig.add_trace(
    go.Bar(
        x=values_gaspacho,
        y=labels_gaspacho,
        marker_color=colors,
        orientation="h",
        hoverinfo="skip"
    ),
    row=2,
    col=1
)
fig.add_trace(
    go.Bar(
        x=values_sandwich,
        y=labels_sandwich,
        marker_color=colors,
        orientation="h",
        hoverinfo="skip"
    ),
    row=3,
    col=1
)
fig.add_trace(
    go.Bar(
        x=values_brownie,
        y=labels_brownie,
        marker_color=colors,
        orientation="h",
        hoverinfo="skip"
    ),
    row=4,
    col=1
)

fig.update_xaxes(
    tickangle=0,
    ticklabelstandoff=5,
    gridcolor="#5B8FCB"
)

fig.update_yaxes(
    tickangle=0,
    ticklabelstandoff=10,
    autorange="reversed"
)

fig.update_layout(
    plot_bgcolor="#FDF2ED",
    paper_bgcolor="#FDF2ED",
    font_family="Montserrat",
    font_color="#5B8FCB",
    showlegend=False,
    barcornerradius=30,
    xaxis1=dict(range=[0, 6.5]),
    xaxis2=dict(range=[0, 6.5]),
    xaxis3=dict(range=[0, 6.5]),
    xaxis4=dict(title="Empreinte carbone (kg CO2eq/kg)", range=[0, 65]),
    margin=dict(
        t=40,
        b=40,
        l=20,
        r=20
    ),
    width=400,
    height=600
)
```

```python
fig.write_html("../figures/recipes_carbon.html", include_plotlyjs="cdn")
```
