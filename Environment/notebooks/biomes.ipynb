```python
# import libraries
import plotly.graph_objects as go
```

```python
# input data to be displayed
values = [6.06, 1.65, 0.28, 0.15, 0.06]
labels = ["Algues marines", "Herbiers marins", "Récifs coraliens", "Mangroves", "Marais salants"]
parents = ["Biomes", "Biomes", "Biomes", "Biomes", "Biomes", "Biomes"]
colors = ["#1D3B6E", "#209490", "#C73175", "#F6A01E", "#5B8FCB"]
```

```python
fig = go.Figure()

fig.add_trace(
    go.Treemap(
        values=values,
        labels=labels,
        parents=parents,
        textinfo="label+value",
        textfont=dict(color="#FDF2ED", size=15),
        root_color="#97C7EC",
        marker_colors=colors
    )
)


fig.update_layout(
    plot_bgcolor="rgba(0,0,0,0)",
    paper_bgcolor="rgba(0,0,0,0)",
    font_family="Montserrat",
    hovermode=False,
    modebar=dict(bgcolor="rgba(0,0,0,0)"),
    margin=dict(
        t=30,
        b=20,
        l=20,
        r=20
    ),
    width=1000,
    height=600
)
```

```python
fig.write_html("../figures/biomes.html", include_plotlyjs="cdn")
```
