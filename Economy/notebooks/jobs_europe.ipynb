```python
# import libraries
import pandas as pd
import plotly.graph_objects as go
```

```python
# make dataframe from data
x=['Emplois<br>directs', 'Emplois<br>indirects', 'Emplois<br>induits', 'Total']
#data_jobs =pd.DataFrame({"x":x, "y":[0, 37, 65, 0], "z":[22,17,11,50], "a": [15, 11, 7, 33] })
data_jobs =pd.DataFrame({"x":x, "y":[0, 0, 0, 0], "z":[22,17,11,50], "a": [15, 11, 7,33] })
```

## Plot potential jobs

### PC version

```python
# plot jobs in Europe
fig8 = go.Figure()

fig8.add_trace(
    go.Bar(
        x=data_jobs["x"],
        y=data_jobs["z"],
        name='Produits cultivés et finis en Europe',
        marker_color='#F6A01E',
        text=data_jobs["z"],
        textfont=dict(color="white",size=10)))

fig8.add_trace(
    go.Bar(
        x=data_jobs["x"],
        y=data_jobs["a"],
        name='Produits cultivés hors Europe, finis en Europe',
        marker_color='#209490',
        text=data_jobs["a"],
        textfont=dict(color="white", size=10)))

fig8.update_xaxes(
    showgrid=False,
    ticklabelstandoff=10,
)

fig8.update_yaxes(
    title=dict(
        text="Nombre d'emplois (milliers d'ETP)",
        font=dict(size=16)
    ),
    gridcolor="#1D3B6E",
    zerolinecolor="#1D3B6E",
    tickvals=[0, 20, 40, 60, 80],
    ticklabelstandoff=10,
)

fig8.update_layout(
    barmode='stack',
    paper_bgcolor="#FDF2EE",
    plot_bgcolor="#FDF2EE",
    font_family="Montserrat",
    font_color="#1D3B6E",
    barcornerradius=5,
    legend=dict(
        orientation="h",
        font_family="Montserrat",
        font_color="#113972",
        bgcolor="#FDF2EE",
        yanchor="top",
        y=-0.15,
        xanchor="center",
        x=0.5
    ),
    margin=dict(
        l=20,
        r=20,
        t=30,
        b=20
    ),
    height=400,
    width=900,
)

fig8.show()
```

```python
fig8.write_html("../figures/jobs_europe_pc.html", include_plotlyjs="cdn")
```
