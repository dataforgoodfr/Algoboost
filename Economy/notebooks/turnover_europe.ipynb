```python
# import libraries
import pandas as pd
import plotly.graph_objects as go
```

```python
# load data
data_turnover = pd.read_csv('../data/Algae-industry-Europe-socioeconomic - Socio-economic data.csv')
```

```python
# preview data
data_turnover.head()
```

```python
# rename some columns of the dataframe
data_turnover = data_turnover.rename(columns={"Organism group": "production_type", \
    "Average available turnover in the last 5 years": "turnover_l5y", \
    "Is Algae considered the main business stream": "is_algae_business", \
    "% of business focused on algae - ESTIMATED": "algae_business_percent_estimated"})

# keep only relevant columns
data_turnover = data_turnover[["ID", "Country", "production_type", "turnover_l5y", "is_algae_business"]]

# keep only relevant data
mask_business = data_turnover['is_algae_business'] == "Yes"
mask_production = data_turnover['production_type'] == "Macroalgae"
data_turnover = data_turnover[mask_business & mask_production]

# format numbers
data_turnover["turnover_l5y"] = data_turnover["turnover_l5y"].str.replace(',', "").fillna(0)
data_turnover["turnover_l5y"] = data_turnover["turnover_l5y"].astype('int64')

# group and sort data
data_turnover = data_turnover.rename(columns={"turnover_l5y": "Chiffre_d_affaires", "Country": "Pays"})
data_turnover = data_turnover[["Pays", "Chiffre_d_affaires"]].groupby(by="Pays").sum().reset_index()
data_turnover = data_turnover.sort_values(by="Chiffre_d_affaires", ascending=False)

# keep only relevant data
mask_turnover = data_turnover["Chiffre_d_affaires"] > 0
data_turnover = data_turnover[mask_turnover]
```

```python
# translate country names
def translate_country(row):
  if (row["Pays"] == "Ireland"):
    return 'Irlande'
  elif (row["Pays"] == "Norway"):
    return 'Norvège'
  elif (row["Pays"] == "Iceland"):
    return 'Islande'
  elif (row["Pays"] == "Spain"):
    return 'Espagne'
  elif (row["Pays"] == "Estonia"):
    return 'Estonie'
  elif (row["Pays"] == "UK"):
    return 'Royaume-Uni'
  elif (row["Pays"] == "Denmark"):
    return 'Danemark'
  elif (row["Pays"] == "Sweden"):
    return 'Suède'
  else:
    return row["Pays"]
data_turnover["Pays"] = data_turnover.apply(translate_country, axis=1)
```

## Plot turnover by country

### PC version

```python
# set colors
colors= ['#0074E4','#58B7B0', '#C73175', '#D85F9F', '#991358','#F6A01E', '#1E52A1', \
    '#E1D3E9', '#13716A','#E5800B', '#3573B9', '#CDE8FA', '#58B7B0', '#D85F9F', '#F9C06B', \
    '#97C7EC', '#E1D3E9', '#F18882', '#FBDDD9', '#C3ADD4', '#F8C2BB']

# plot turnover
fig7 = go.Figure()

fig7.add_trace(
    go.Bar(
        x=data_turnover["Pays"],
        y=data_turnover["Chiffre_d_affaires"],
        marker_color=colors,
        hovertemplate="""<span style='font-family:Montserrat; font-weight:600; color:#FDF2EE'>Pays : %{x}</span><br>""" + \
            """<span style='font-family:Montserrat; font-weight:600; color:#FDF2EE'>Chiffre d'affaires : %{y}</span>""" + \
            """<extra></extra>""",
    ))
    
fig7.update_xaxes(
    showgrid=False,
    ticklabelstandoff=10,
)

fig7.update_yaxes(
    title=dict(
        text="Chiffre d'affaires (euros)",
        font=dict(size=16)
    ),
    zerolinecolor="#1D3B6E",
    gridcolor="#1D3B6E",
    tickvals=[0, 10000000, 20000000, 30000000, 40000000, 50000000],
    ticklabelstandoff=10,
)

fig7.update_layout(
    font_family="Montserrat",
    font_color="#1D3B6E",
    barcornerradius=15,
    paper_bgcolor="#F9BF6B",
    plot_bgcolor="#F9BF6B",
    margin=dict(
        l=20,
        r=20,
        t=30,
        b=20
    ),
    height=400,
    width=900,
)
fig7.show()
```

```python
fig7.write_html("../figures/turnover_europe_pc.html", include_plotlyjs="cdn")
```
