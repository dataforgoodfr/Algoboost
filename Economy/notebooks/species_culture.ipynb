```python
# import libraries
import pandas as pd
import plotly.graph_objects as go
```

```python
# get data of species procuded by culture in Europe 
list_algae_culture = ['Alaria sp.','Ascophyllum nodosum','Asparagopsis sp.','Brown seaweed','Calliblepharis jubata','Caulerpa sp.','Chondrus sp.','Codium sp.','Falkenbergia sp.','Fucus sp.','Gracilaria sp.','Gracilariopsis longissima','Himanthalia sp.','Kappaphycus','Laminaria sp.','Palmaria sp.','Porphyra sp.','Saccharina sp.','Schizymenia jonssonii','Ulva sp.','Ulvella lens','Undaria sp.','Espèce inconnue','Vertebrata lanosa']
list_nb_culture = [17,1,1,2,1,1,2,2,1,5,3,2,2,1,10,8,2,29,1,15,1,3,3,1]

# make dataframe
df_culture = pd.DataFrame({"algae":list_algae_culture, "nb_culture": list_nb_culture})
df_culture_sort = df_culture.sort_values(by="nb_culture", ascending=False)
```

## Plot species produced by culture in Europe

### PC version

```python
# set colors
colors_culture = ['#F6A01E','#E5800B', '#F18882', '#1D3B6E', '#5B8FCB','#209490' , '#3573B9', \
    '#13716A','#F9C06B', '#1E52A1','#D85F9F', '#C73175', '#58B7B0', '#991358', '#CDE8FA', \
    '#97C7EC'  , '#97C7EC', '#E1D3E9', '#F18882', '#FBDDD9', '#C3ADD4', 'F8C2BB']
font_colors = ['#FDF2EE','#FDF2EE', '#FDF2EE', '#FDF2EE', '#FDF2EE','#FDF2EE' , '#FDF2EE', \
    '#FDF2EE','#113972', '#FDF2EE','#FDF2EE', '#FDF2EE', '#FDF2EE', '#FDF2EE', '#113972', \
    '#FDF2EE'  , '#FDF2EE', '#FDF2EE', '#FDF2EE', '#FDF2EE', '#FDF2EE', '#FDF2EE']

# plot production
fig5 = go.Figure()

fig5.add_trace(
    go.Treemap(
        labels=df_culture_sort["algae"],
        parents=["",] * 15,
        values=df_culture_sort["nb_culture"],
        texttemplate="""<span style="font-weight:800">%{label}</span> <br>""" + \
        """%{value} entreprises (%{percentEntry:.1%})""",
        marker_colors=colors_culture,
        textfont_color= font_colors,
        root_color="#FDF2EE"
        ))

fig5.update_traces(
    marker=dict(cornerradius=5),
    hoverinfo="skip")

fig5.update_layout(
    paper_bgcolor ="#FDF2EE",
    plot_bgcolor ="#FDF2EE",
    font_family="Montserrat",
    margin=dict(
        l=10,
        r=10,
        t=10,
        b=10
    ),
    height=400,
    width=900,
  )

fig5.show()
```

```python
fig5.write_html("../figures/species_culture_pc.html", include_plotlyjs="cdn")
```
