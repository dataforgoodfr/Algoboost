```python
# import libraries
import pandas as pd
import plotly.graph_objects as go
```

```python
# load data
data = pd.read_csv("../data/vitamins.csv", sep=";")
```

```python
# preview data
data.head()
```

```python
# drop unused data
mask_keep = data["Inclusion"] == 1
data = data.loc[mask_keep, :].reset_index(drop=True)
```

```python
# format numbers from strings to float
columns = data.columns[5:]
for column in columns:
    mask_ok = data[column].notnull()
    data.loc[mask_ok, column] = [strnb.replace(",", ".") for strnb in data.loc[mask_ok, column]]
#data[columns] = data[columns].astype("float")
```

```python
# format nans
mask_nan = data.isnull()
data[mask_nan] = "nd"
```

```python
# create vitamin description
vitamin_info = [
    "Recommandation<br>journalière (mg/jour)<br>Femmes : 90<br>Hommes : 90",
    "Recommandation<br>journalière (mg/jour)<br>Femmes : 9<br>Hommes : 10",
    "Recommandation<br>journalière (mg EN/MJ)<br>Femmes : 1,3<br>Hommes : 1,3",
    "Recommandation<br>journalière (mg/jour)<br>Femmes : 1,3<br>Hommes : 1,3",
    "Recommandation<br>journalière (µg/jour)<br>Femmes : 79<br>Hommes : 79",
    "Recommandation<br>journalière (mg/jour)<br>Femmes : 1,3<br>Hommes : 1,5",
    "Recommandation<br>journalière (mg/MJ)<br>Femmes : 0,072<br>Hommes : 0,072",
    "Recommandation<br>journalière (mg/jour)<br>Femmes : 5<br>Hommes : 6",
    "Recommandation<br>journalière (µg/jour)<br>Femmes : 490<br>Hommes : 580",
    "Recommandation<br>journalière (µg EFA/jour)<br>Femmes : 250<br>Hommes : 250",
    "Recommandation<br>journalière (µg/jour)<br>Femmes : 40<br>Hommes : 40",
    "Recommandation<br>journalière (µg/jour)<br>Femmes : 4<br>Hommes : 4",
    "Recommandation<br>journalière (µg/jour)<br>Femmes : 15<br>Hommes : 15"
]
```

```python
fig = go.Figure()
config = {'displayModeBar': True}

customscale = [[ 0.0, "#CCE9FF"], [0.05, "#93C9F9"], [1.0, "#007DED"]]

fig.add_trace(
    go.Heatmap(
        z=data[columns],
        x="<b>" + columns + "</b><br>" + vitamin_info,
        y="""<span style="color:""" + data["Couleur"] + """"><b>""" + \
            data["Nom vernaculaire"] + "</b><br><i>" + data["Nom scientifique"] + """</i></span>""",

        text=data[columns].astype(str),
        texttemplate="%{text}",
        textfont=dict(size=10, color="#113972"),
        colorscale=customscale,
        showscale=False,
        hoverinfo="skip"
    ),
)

fig.update_xaxes(
    showgrid=False,
    side="top",
    tickangle=0,
    ticklabelstandoff=10,
    color="#007DED"
),

fig.update_yaxes(
    showgrid=False,
    tickangle=0,
    ticklabelstandoff=10,
    autorange="reversed"
)

fig.update_layout(
    plot_bgcolor="white",
    font_family="Montserrat",
    margin=dict(
        t=20,
        b=20,
        l=20,
        r=20
    ),
    modebar=dict(
        orientation="h",
    ),
    height=800,
    width=2100
)

fig.show(config=config)
```

```python
fig.write_html("../figures/vitamins.html", include_plotlyjs="cdn")
```
